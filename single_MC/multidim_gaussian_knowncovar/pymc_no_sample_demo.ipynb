```python
import os
import numpy as np
import pymc as pm
import pytensor
if os.environ.get("SIMPLEVI_PYTENSOR_CXX"):
    pytensor.config.cxx = os.environ["SIMPLEVI_PYTENSOR_CXX"]
```

## we demonstrate what happens when we have no observations and try to run conj multid gaussian VI in pymc

```python
observed_data_1 = []
observed_data_2 = None

try:
    with pm.Model() as first_model:
        # Define priors
        mu = pm.Normal('mu', np.zeros(40), np.ones(40) * 10, shape=40)
        # Define likelihood with 3 observations
        y = pm.Normal('y', mu, sigma=np.ones(40), observed=observed_data_1)

        advi = pm.ADVI(random_seed=42)
        single_approx = advi.fit()
except Exception as e:
    print("Fails with empty list:", e)
```

```python

try:
    with pm.Model() as second_model:
        # Define priors
        mu = pm.Normal('mu', np.zeros(40), np.ones(40) * 10, shape=40)
        # Define likelihood with 3 observations
        y = pm.Normal('y', mu, sigma=np.ones(40), observed=observed_data_2)

        advi = pm.ADVI(random_seed=42)
        single_approx = advi.fit()
except Exception as e:
    print("Fails with None:", e)
```
