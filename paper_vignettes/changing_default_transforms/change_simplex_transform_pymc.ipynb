# Changing the simplex transform in PyMC

ADVI maps each parameter to an unconstrained space with a transform. For `pm.Dirichlet`, PyMC's default
is a softmax of zero-sum logits. To use a different transform, pass it to the prior's `default_transform`
argument.

PyMC has no built-in stick-breaking simplex transform, so we use the user-defined
`StickBreakingSimplexTransform` from `modulars/pymc_rr_test.py` (written to match NumPyro's default simplex transform).

```python
import numpy as np
import pymc as pm

from modulars.pymc_rr_test import StickBreakingSimplexTransform

alpha = np.ones(3)
```

Default transform:

```python
with pm.Model() as default_model:
    theta = pm.Dirichlet("theta", a=alpha)

default_model.rvs_to_transforms
```

Changed transform: the only difference is `default_transform=...`.

```python
with pm.Model() as stickbreaking_model:
    theta = pm.Dirichlet("theta", a=alpha, default_transform=StickBreakingSimplexTransform())

stickbreaking_model.rvs_to_transforms
```
